```python
import numpy as np
rnd = np.random.default_rng()
a = rnd.integers(0, 10, 6, endpoint=True)
b = rnd.integers(0, 10, 6, endpoint=True)
print("Исходное а:", a)
print("Исходное b:", b)

common_vals = np.intersect1d(a, b)

c = common_vals.copy()

for v in common_vals:
    a[a == v] = 0
    b[b == v] = 0

print("a:", a)
print("b:", b)
print("c (уникальные общие):", c)
```
